```python
import pandas as pd
import numpy as np

from google.colab import files

# Upload dataset
uploaded = files.upload()

# Read uploaded CSV
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

print("Dataset loaded successfully!")
print(df.head())
```

```python
# ==========================================
# TASK 1: DATA CLEANING & PREPROCESSING
# ==========================================

# 1. Inspect dataset structure

print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())

# 2. Identify missing values

print("\nMissing Values:")
print(df.isnull().sum())

# 3. Identify duplicate records

print("\nDuplicate Records:")
print(df.duplicated().sum())

# 4. Display statistical summary

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# ==========================================
# DATA CLEANING
# ==========================================

# Make a copy of the dataset
cleaned_df = df.copy()

# 1. Remove duplicate records

cleaned_df = cleaned_df.drop_duplicates()

# 2. Handle missing values

for column in cleaned_df.columns:

    if cleaned_df[column].dtype == "object":

        cleaned_df[column] = cleaned_df[column].fillna(
            cleaned_df[column].mode()[0]
        )

    else:

        cleaned_df[column] = cleaned_df[column].fillna(
            cleaned_df[column].median()
        )

# 3. Remove extra spaces from text columns

for column in cleaned_df.select_dtypes(include="object").columns:

    cleaned_df[column] = cleaned_df[column].str.strip()

# 4. Display cleaned dataset

print("Cleaned Dataset:")
display(cleaned_df.head())

# 5. Check missing values after cleaning

print("\nMissing Values After Cleaning:")
print(cleaned_df.isnull().sum())

# 6. Check duplicate records after cleaning

print("\nDuplicate Records After Cleaning:")
print(cleaned_df.duplicated().sum())
```

```python
# ==========================================
# COMPLETE DATA CLEANING
# ==========================================

# Create a copy of the original dataset
cleaned_df = df.copy()

# 1. Remove duplicate records
cleaned_df = cleaned_df.drop_duplicates()

# 2. Remove extra spaces from text columns
for column in cleaned_df.select_dtypes(include="object").columns:
    cleaned_df[column] = cleaned_df[column].str.strip()

# 3. Standardize Gender values
cleaned_df["Gender"] = (
    cleaned_df["Gender"]
    .astype("string")
    .str.lower()
    .replace({
        "m": "Male",
        "male": "Male",
        "f": "Female",
        "female": "Female"
    })
)

# 4. Standardize City values
cleaned_df["City"] = (
    cleaned_df["City"]
    .astype("string")
    .str.lower()
    .replace({
        "bangalore": "Bengaluru",
        "mysore": "Mysuru"
    })
)

# 5. Convert Age and Score to numeric
cleaned_df["Age"] = pd.to_numeric(
    cleaned_df["Age"], errors="coerce"
)

cleaned_df["Score"] = pd.to_numeric(
    cleaned_df["Score"], errors="coerce"
)

# 6. Handle missing values
cleaned_df["Age"] = cleaned_df["Age"].fillna(
    cleaned_df["Age"].median()
)

cleaned_df["Score"] = cleaned_df["Score"].fillna(
    cleaned_df["Score"].median()
)

cleaned_df["Gender"] = cleaned_df["Gender"].fillna(
    cleaned_df["Gender"].mode()[0]
)

cleaned_df["City"] = cleaned_df["City"].fillna(
    cleaned_df["City"].mode()[0]
)

# 7. Display cleaned dataset
print("CLEANED DATASET")
display(cleaned_df)

# 8. Check remaining missing values
print("\nMissing Values After Cleaning:")
print(cleaned_df.isnull().sum())

# 9. Check duplicate records
print("\nDuplicate Records After Cleaning:")
print(cleaned_df.duplicated().sum())
```

```python
# Save the final cleaned dataset

cleaned_df.to_csv("cleaned_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
