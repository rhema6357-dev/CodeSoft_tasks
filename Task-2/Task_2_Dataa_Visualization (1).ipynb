```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

%matplotlib inline

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
import pandas as pd

df = pd.read_csv("Sample-Superstore.csv")

df.head()
```

```python
# Check the number of rows and columns
print("Dataset Shape:")
print(df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check data types and missing values
print("\nDataset Information:")
df.info()
```

```python
# Check missing values
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
# Check duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
# Convert date columns to datetime format

df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])

# Check the data types
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
# Create Year and Month columns for analysis

df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()

# Display the new columns
df[["Order Date", "Year", "Month", "Month Name"]].head()
```

```python
# Calculate profit margin

df["Profit Margin"] = (df["Profit"] / df["Sales"]) * 100

df[["Sales", "Profit", "Profit Margin"]].head()
```

```python
# Display the final dataset information

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

df.head()
```

```python
# ==========================================
# BAR CHART - SALES BY CATEGORY
# ==========================================

category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))

plt.bar(
    category_sales.index,
    category_sales.values
)

plt.title("Total Sales by Category", fontsize=16, fontweight="bold")
plt.xlabel("Product Category", fontsize=12)
plt.ylabel("Total Sales", fontsize=12)

plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

### Insight

The bar chart compares the total sales generated by each product category.
It helps identify which categories contribute the most to overall sales
and which categories have relatively lower sales.

```python
# ==========================================
# LINE CHART - MONTHLY SALES TREND
# ==========================================

monthly_sales = (
    df.groupby(df["Order Date"].dt.to_period("M"))["Sales"]
    .sum()
)

monthly_sales.index = monthly_sales.index.astype(str)

plt.figure(figsize=(12, 6))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    linewidth=2
)

plt.title("Monthly Sales Trend", fontsize=16, fontweight="bold")
plt.xlabel("Month", fontsize=12)
plt.ylabel("Total Sales", fontsize=12)

plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# 1. BAR CHART - SALES BY CATEGORY
# ==========================================

category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))

bars = plt.bar(category_sales.index, category_sales.values)

plt.title("Total Sales by Category", fontsize=16, fontweight="bold")
plt.xlabel("Product Category", fontsize=12)
plt.ylabel("Total Sales", fontsize=12)

plt.grid(axis="y", alpha=0.3)

# Add values above bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f"${height:,.0f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# 3. PIE CHART - SALES BY REGION
# ==========================================

region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8, 8))

plt.pie(
    region_sales.values,
    labels=region_sales.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title(
    "Sales Distribution by Region",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# 4. HISTOGRAM - SALES DISTRIBUTION
# ==========================================

plt.figure(figsize=(10, 6))

plt.hist(
    df["Sales"],
    bins=30,
    edgecolor="black"
)

plt.title(
    "Distribution of Sales",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Sales", fontsize=12)
plt.ylabel("Frequency", fontsize=12)

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

### Insight

The histogram shows the distribution of individual sales values. It helps
identify the most common sales ranges and reveals whether the data contains
very high or very low sales values.

```python
# ==========================================
# 5. SCATTER PLOT - SALES VS PROFIT
# ==========================================

plt.figure(figsize=(10, 6))

plt.scatter(
    df["Sales"],
    df["Profit"],
    alpha=0.5
)

plt.title(
    "Sales vs Profit",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Sales", fontsize=12)
plt.ylabel("Profit", fontsize=12)

plt.axhline(y=0, linestyle="--", linewidth=1)

plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### Insight

The scatter plot examines the relationship between sales and profit.
The distribution of points helps identify whether higher sales are generally
associated with higher profit and also highlights transactions with low or
negative profit.

# Key Performance Indicators

The following metrics provide a high-level summary of the business performance.

```python
# ==========================================
# KEY PERFORMANCE INDICATORS
# ==========================================

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_quantity = df["Quantity"].sum()
total_orders = df["Order ID"].nunique()
average_sales = df["Sales"].mean()

print("=" * 45)
print("          BUSINESS PERFORMANCE")
print("=" * 45)

print(f"Total Sales       : ${total_sales:,.2f}")
print(f"Total Profit      : ${total_profit:,.2f}")
print(f"Total Quantity    : {total_quantity:,}")
print(f"Total Orders      : {total_orders:,}")
print(f"Average Sale      : ${average_sales:,.2f}")

print("=" * 45)
```

## Profit Analysis by Category

```python
# ==========================================
# PROFIT BY CATEGORY
# ==========================================

category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

bars = plt.bar(
    category_profit.index,
    category_profit.values
)

plt.title(
    "Total Profit by Category",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Category")
plt.ylabel("Profit")

plt.grid(axis="y", alpha=0.3)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f"${height:,.0f}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CORRELATION ANALYSIS
# ==========================================

correlation = df[[
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]].corr()

print(correlation)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title(
    "Correlation Between Numerical Variables",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()
plt.show()
```

# Interactive Dashboard

An interactive dashboard is created using Plotly to allow users to explore
sales performance across categories and regions.

```python
# ==========================================
# INTERACTIVE SALES DASHBOARD
# ==========================================

fig = px.scatter(
    df,
    x="Sales",
    y="Profit",
    color="Category",
    size="Quantity",
    hover_data=[
        "Region",
        "Sub-Category",
        "Product Name"
    ],
    title="Interactive Sales vs Profit Dashboard"
)

fig.update_layout(
    width=1000,
    height=600
)

fig.show()
```

```python
# ==========================================
# INTERACTIVE SALES BY REGION
# ==========================================

fig = px.bar(
    region_sales.reset_index(),
    x="Region",
    y="Sales",
    color="Region",
    title="Interactive Sales by Region"
)

fig.update_layout(
    width=900,
    height=500
)

fig.show()
```
